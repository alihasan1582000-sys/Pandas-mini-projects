The managers for the "Laptop", "Audio", and "Tablet" departments each want their own report.
Task: Write a Python for loop that iterates through the unique categories in your dataset. (Hint: master_df['Category'].unique() gives you a list of them).
Inside the loop, filter the master_df to create a temporary DataFrame for just that category.

```python
#Toy dataset for Practice 
# Table 1: Customer Database
customers = pd.DataFrame({
    'Customer_ID': [1, 2, 3, 4, 5],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Evan'],
    'VIP_Member': [True, False, False, True, False]
})

# Table 2: Product Database (Note the missing price for P03!)
products = pd.DataFrame({
    'Product_ID': ['P01', 'P02', 'P03', 'P04'],
    'Category': ['Laptop', 'Audio', 'Audio', 'Tablet'],
    'Price': [1200.0, 150.0, np.nan, 300.0] 
})

# Table 3: Order History
orders = pd.DataFrame({
    'Order_ID': [1001, 1002, 1003, 1004, 1005, 1006],
    'Customer_ID': [1, 2, 1, 4, 99, 3], # Customer 99 doesn't exist in our DB!
    'Product_ID': ['P01', 'P03', 'P02', 'P04', 'P01', 'P03'],
    'Quantity': [1, 2, 1, 3, 1, 1]
})
```

```python
#CleanUP
products['Price']=products['Price'].fillna(200.0)
```

```python
products
```

```python
#merge of the three tables
master_df=orders.merge(customers, on='Customer_ID', how='left')
master_df=master_df.merge(products, on='Product_ID', how='left')
```

```python
master_df['Total_Revenue']=master_df['Quantity']*master_df['Price']
```

```python
master_df
```

```python
for category in master_df['Category'].dropna().unique():
    category_df=master_df[master_df['Category']==category]
```

```python
summ_table=category_df.groupby('VIP_Member').agg({
    'Total_Revenue': 'sum',
    'Quantity': 'sum'
}).reset_index()
```
